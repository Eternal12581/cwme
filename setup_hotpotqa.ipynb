```python
import random
import json
```

```python
with open('./hotpot_dev_distractor_v1.json', 'r') as f:
    hotpot_test = json.load(f)
    
# sample 200 random questions and save into new json file
random_sample = random.sample(range(0, 7405), 200)
new_hotpot = [hotpot_test[i] for i in random_sample]
len(new_hotpot)
with open('./hotpot_custom.json', 'w') as f:
    json.dump(new_hotpot, f, indent=4)
```
